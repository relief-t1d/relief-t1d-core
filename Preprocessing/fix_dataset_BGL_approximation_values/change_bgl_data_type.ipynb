# Diatrend

```python
df_diatrend_original = pd.read_parquet('c:/Users/Ciro/C/_UGR_PDI/RELIEF-T1D/Datasets/2025-12-05/Preprocessed/diatrend/Final_versions_V4/Filtered_for_prediction/Glucose_measurements_FILTERED_2025-11-20.parquet')
df_diatrend_original
```

```python
df_diatrend_original.info()
```

## Round Measurement to 0 decimals and change Patient_ID to string

```python
df_diatrend_round_0_id_str_measurement_int = df_diatrend_original.copy()
df_diatrend_round_0_id_str_measurement_int['Patient_ID'] = df_diatrend_round_0_id_str_measurement_int['Patient_ID'].astype(str)
df_diatrend_round_0_id_str_measurement_int['Measurement'] = df_diatrend_round_0_id_str_measurement_int['Measurement'].round(0).astype('Int16')
df_diatrend_round_0_id_str_measurement_int
```

## Lowercase column names

```python
df_diatrend_round_0_id_str_measurement_int.columns = df_diatrend_round_0_id_str_measurement_int.columns.str.lower()
df_diatrend_round_0_id_str_measurement_int                                                                                                
```

```python
df_diatrend_round_0_id_str_measurement_int.info()
```

## Save as parquet  

```python
df_diatrend_round_0_id_str_measurement_int.to_parquet('c:/Users/Ciro/C/_UGR_PDI/RELIEF-T1D/Datasets/2025-12-05/Preprocessed/diatrend/Final_versions_V4/Filtered_for_prediction/Glucose_measurements_DiaTrend_FILTERED_2026-03-27.parquet')
```

The changes made to the original DataFrame include:
1. **Patient_ID**: The `Patient_ID` column was converted from its original data type (likely an integer) to a string. This allows for better handling of patient identifiers, especially if they are not purely numeric or if leading zeros are important.
2. **Measurement**: The `Measurement` column was rounded to 0 decimal places and converted to an integer type (`Int16`). 
3. **Column Names**: The column names were converted to lowercase.
