Get the stats from the folds of the cross validation

```python
import os
import pandas as pd
import numpy as np
from scipy.stats import t
```

```python
def list_csv_files(directory):
    csv_files = [file for file in os.listdir(directory) if file.endswith('.csv')]
    return csv_files

directory_1 = 'c:/Users/Ciro/Mi Unidad (PDI)/_UGR_PDI_2024-05-30/RELIEF-T1D/Python_projects/CompareResults/DataPool/CrossValidation/Exp3/'
csv_files_1 = list_csv_files(directory_1)


def merge_csv_files(directory):
    csv_files = list_csv_files(directory)
    merged_df = pd.DataFrame()

    for file in csv_files:
        file_path = os.path.join(directory, file)
        df = pd.read_csv(file_path)
        df['file_name'] = file
        merged_df = pd.concat([merged_df, df], ignore_index=True)
    
    return merged_df



df_new_result_1 = merge_csv_files(directory_1)
df_new_result_1 = df_new_result_1.round(2)
df_new_result_1
```

```python
df_new_result_1['Algorithm'] = df_new_result_1['file_name'].apply(lambda x: x.split('_')[-3])
df_new_result_1['PH'] = df_new_result_1['file_name'].apply(lambda x: x.split('_')[-2].split('.')[0][1:])
df_new_result_1['PH'] = df_new_result_1['PH'].astype(int)
df_new_result_1['Fold'] = df_new_result_1['file_name'].apply(lambda x: x.split('_')[-1].split('.')[0][-1])
df_new_result_1['Fold'] = df_new_result_1['Fold'].astype(int)
df_new_result_1
```

```python
df_new_result_1 = df_new_result_1.drop(columns=['file_name'])
df_new_result_1
```

```python
desired_order = ["ENTIRE", "TBR_2", "TBR_1", "TIR", "TAR_1", "TAR_2"]

df_new_result_1["Range"] = pd.Categorical(df_new_result_1["Range"], categories=desired_order, ordered=True)
df_new_result_1
```

```python
def summarize_metrics(df: pd.DataFrame, group_cols: list, metric_cols: list) -> pd.DataFrame:
    """
    Compute summary statistics for metrics over cross-validation folds.

    Parameters
    ----------
    df : pd.DataFrame
        Your full result table (must include columns like 'Fold', 'RMSE', etc.).
    group_cols : list
        Columns to group by (e.g. ['Algorithm', 'PH', 'Range']).
    metric_cols : list
        Metrics to summarize (e.g. ['RMSE', 'MAE', 'MAPE']).

    Returns
    -------
    pd.DataFrame
        One row per group with full summary stats: mean, std, IC95%, median, IQR, CV.
    """
    results = []

    for keys, group in df.groupby(group_cols, observed=True):
        row = dict(zip(group_cols, keys)) if isinstance(keys, tuple) else {group_cols[0]: keys}
        n = len(group)

        for metric in metric_cols:
            values = group[metric].dropna()
            mean = values.mean()
            std = values.std(ddof=1)
            median = values.median()
            min = values.min()
            max = values.max()
            q1 = values.quantile(0.25)
            q3 = values.quantile(0.75)
            iqr = q3 - q1
            cv = std / mean if mean != 0 else np.nan

            # 95% confidence interval
            t_val = t.ppf(0.975, df=n-1) if n > 1 else np.nan
            margin = t_val * std / np.sqrt(n) if n > 1 else np.nan
            ic_lower = mean - margin if n > 1 else np.nan
            ic_upper = mean + margin if n > 1 else np.nan

            row.update({
                f"{metric}_mean": mean,
                f"{metric}_std": std,
                f"{metric}_cv": cv,
                f"{metric}_median": median,
                f"{metric}_min": min,
                f"{metric}_iq_25": q1,
                f"{metric}_iq_75": q3,
                f"{metric}_max": max,
                f"{metric}_iqr": iqr,
                f"{metric}_ci95_lower": ic_lower,
                f"{metric}_ci95_upper": ic_upper,
            })

        results.append(row)

    return pd.DataFrame(results)
```

```python
group_cols = ['Algorithm', 'PH', 'Range']
metric_cols = ['A + B', 'RMSE']
df_summary_1 = summarize_metrics(df_new_result_1, group_cols, metric_cols)
df_summary_1 = df_summary_1.round(2)
df_summary_1
```
