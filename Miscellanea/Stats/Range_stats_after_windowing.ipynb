# DiaTrend

```python
df_windows_horizon_4_diatrend = pd.read_parquet('c:/Users/Ciro/C/_UGR_PDI/RELIEF-T1D/Datasets/2025-12-05/Preprocessed/diatrend/files_for_crossvalidation/windows/windows_horizon_4.parquet')
df_windows_horizon_4_diatrend
```

## Patient number

```python
df_windows_horizon_4_diatrend['patient_id'].unique().shape
```

```python
target_diatrend = df_windows_horizon_4_diatrend['y']
target_diatrend
```

```python
target_diatrend.describe().round(2)
```

## Range array creation

```python
df_target_vector_diatrend_TBR_2 = target_diatrend[target_diatrend < 54]

df_target_vector_diatrend_TBR_1 = target_diatrend[(target_diatrend >= 54) & (target_diatrend < 70)]

df_target_vector_diatrend_TIR = target_diatrend[(target_diatrend >= 70) & (target_diatrend < 181)]

df_target_vector_diatrend_TAR_1 = target_diatrend[(target_diatrend >= 181) & (target_diatrend < 251)]

df_target_vector_diatrend_TAR_2 = target_diatrend[(target_diatrend >= 251)]
```

## Dataframe with stats 
Count, % Total, mean, std

```python
df_stats_diatrend = pd.DataFrame(columns=['ENTIRE', 'TBR_2', 'TBR_1', 'TIR', 'TAR_1', 'TAR_2'], index=['Count', '% Total', 'Mean', 'Std'])
df_stats_diatrend.index.name = 'Stats'
df_stats_diatrend
```

```python
# Fill the dataframe with the stats for each range with for loop
ranges = ['ENTIRE', 'TBR_2', 'TBR_1', 'TIR', 'TAR_1', 'TAR_2']
for range in ranges:
    if range == 'ENTIRE':
        data = target_diatrend
    else:
        data = eval(f'df_target_vector_diatrend_{range}')
    
    df_stats_diatrend.loc['Count', range] = data.shape[0]
    df_stats_diatrend.loc['% Total', range] = round((data.shape[0] / target_diatrend.shape[0]) * 100, 2)
    df_stats_diatrend.loc['Mean', range] = data.mean().round(2)
    df_stats_diatrend.loc['Std', range] = data.std().round(2)
    
df_stats_diatrend
```

```python
# Transform dataframe to have percentaje of total and total in the same cell and mean and std in the same cell
df_stats_diatrend_combined = pd.DataFrame(columns=['ENTIRE', 'TBR_2', 'TBR_1', 'TIR', 'TAR_1', 'TAR_2'], index=['Count (%)', 'Mean ±Std'])
df_stats_diatrend_combined.index.name = 'Stats'
for range in ranges:
    count = df_stats_diatrend.loc['Count', range]
    percent_total = df_stats_diatrend.loc['% Total', range]
    mean = df_stats_diatrend.loc['Mean', range]
    std = df_stats_diatrend.loc['Std', range]
    
    df_stats_diatrend_combined.loc['Count (%)', range] = f'{count} ({percent_total}%)'
    df_stats_diatrend_combined.loc['Mean ±Std', range] = f'{mean} ±{std}'

df_stats_diatrend_combined
```

```python
# Traspose the dataframe to have the ranges as index and stats as columns
df_stats_diatrend_combined_transposed = df_stats_diatrend_combined.T
df_stats_diatrend_combined_transposed
```
