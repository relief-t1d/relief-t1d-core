### Data distribution of the glucose levels measurements

#### Plot

```python
darkgray_square = dict(markerfacecolor='darkgray', marker='s')

fig, ax = plt.subplots(figsize=(10,2))

bp_with_fliers = ax.boxplot(df_glucose_measurements_ALL['y'], vert=False, flierprops=darkgray_square, widths=0.5, showfliers=True)

for median in bp_with_fliers['medians']:
    median.set_color('darkgray')

x_ticks_values = np.arange(0, 510, 100)
x_ticks_labels = list(map(str, x_ticks_values))
x_ticks = ax.xaxis.set_ticks(x_ticks_values)
x_labels = ax.xaxis.set_ticklabels(x_ticks_labels)

y_labels = ax.yaxis.set_ticklabels(['All'])

ax.grid(axis='x', linestyle=':', linewidth='0.5', color='black')

fig.savefig('attd/T1DGranada_ALL.png', dpi=900, bbox_inches='tight')
```

### Data distribution of the glucose levels measurements by sex

#### Box plot

```python
sex_plot_data = [df_glucose_measurements_male['y'], df_glucose_measurements_female['y']]
darkgray_square = dict(markerfacecolor='darkgray', marker='s')

fig, ax = plt.subplots(figsize=(10,3))

bp_with_fliers = ax.boxplot(sex_plot_data, vert=False, flierprops=darkgray_square, widths=0.5)

for median in bp_with_fliers['medians']:
    median.set_color('darkgray')

x_ticks_values = np.arange(0, 510, 100)
x_ticks_labels = list(map(str, x_ticks_values))
x_ticks = ax.xaxis.set_ticks(x_ticks_values)
x_labels = ax.xaxis.set_ticklabels(x_ticks_labels)

y_labels = ax.yaxis.set_ticklabels(['Male', 'Female'])

ax.grid(axis='x', linestyle=':', linewidth='0.5', color='black')

fig.savefig('attd/T1DGranada_sex.png', dpi=900, bbox_inches='tight')
```

### Data distribution of the glucose levels measurements by age

#### Boxplot

```python
age_vs_measurement_plot_data = [df_glucose_measurements_young['y'], df_glucose_measurements_youngAdult['y'], df_glucose_measurements_adult['y'], df_glucose_measurements_elder['y']]
darkgray_square = dict(markerfacecolor='darkgray', marker='s')

fig, ax = plt.subplots(figsize=(10,6))

bp_with_fliers = ax.boxplot(age_vs_measurement_plot_data, vert=False, flierprops=darkgray_square, widths=0.5)

for median in bp_with_fliers['medians']:
    median.set_color('darkgray')

x_ticks_values = np.arange(0, 510, 100)
x_ticks_labels = list(map(str, x_ticks_values))
x_ticks = ax.xaxis.set_ticks(x_ticks_values)
x_labels = ax.xaxis.set_ticklabels(x_ticks_labels)

y_labels = ax.yaxis.set_ticklabels(['Age < 31', '31 <= Age < 46', '46 <= Age < 65', 'Age >= 66'])

ax.grid(axis='x', linestyle=':', linewidth='0.5', color='black')

# x_title = ax.set_xlabel('(c)')

# fig.savefig('age_vs_measurements.jpg', dpi=900, bbox_inches='tight')
fig.savefig('attd/T1DGranada_age.png', dpi=900, bbox_inches='tight')
```
