```python
import pandas as pd
import os
```

```python
def list_csv_files(directory):
    csv_files = [file for file in os.listdir(directory) if file.endswith('.csv')]
    return csv_files

all_directory = 'c:/Users/Ciro/C/_UGR_PDI/RELIEF-T1D/Glucose_prediction_2026/Compare_Experiments/2026-01-23/2/Diatrend_fix_BGL_limit/csv_performance/report/'
all_csv_files = list_csv_files(all_directory)

def merge_csv_files(directory):
    csv_files = list_csv_files(directory)
    merged_df = pd.DataFrame()

    for file in csv_files:
        file_path = os.path.join(directory, file)
        df = pd.read_csv(file_path)
        df['file_name'] = file
        merged_df = pd.concat([merged_df, df], ignore_index=True)
    
    return merged_df



df_results = merge_csv_files(all_directory)
df_results = df_results.round(2)

df_results
```

```python
df_results['Algorithm'] = df_results['file_name'].apply(lambda x: x.split('_')[-6])
df_results['PH'] = df_results['file_name'].apply(lambda x: x.split('_')[-5][1:])
df_results['PH'] = df_results['PH'].astype(int)
df_all_results = df_results.sort_values(by=['Algorithm', 'PH'])
df_results = df_results.drop(columns=['file_name'])
df_results = df_results[['Algorithm', 'PH', 'Range', 'A', 'B', 'C', 'D', 'E', 'A + B', 'RMSE', 'MSE', 'MAE', 'MAPE']]
df_results
```
