```python
patient_ids_dedup_v1_train_fold_1 = df_dedup_v1[df_dedup_v1['fold_2'] == 'train']['patient_id'].unique()
patient_ids_dedup_v2_train_fold_1 = df_dedup_v2[df_dedup_v2['fold_2'] == 'train']['patient_id'].unique()
# Check if the patient IDs in the train fold of both DataFrames are the same
patient_ids_dedup_v1_train_fold_1.sort() == patient_ids_dedup_v2_train_fold_1.sort()
```

```python
pd.DataFrame(data={'v1':patient_ids_dedup_v1_train_fold_1, 'v2':patient_ids_dedup_v2_train_fold_1})
```

```python
used_columns= [f"x{i}" for i in range(8)] + ["y"] # Columns to be used for training, validation, and testing
for current_fold in range(0, 5):
    print(f"Fold {current_fold}")
    
    # Filter the DataFrame for training for the current fold
    df_train = df_dedup_v1[df_dedup_v1[f'fold_{current_fold}'] == 'train']
    df_train.reset_index(drop=True, inplace=True)
    df_train = df_train[used_columns] # Select only the used columns
    
    # Filter the DataFrame for validation for the current fold
    df_val = df_dedup_v1[df_dedup_v1[f'fold_{current_fold}'] == 'val']
    df_val.reset_index(drop=True, inplace=True)
    df_val = df_val[used_columns] # Select only the used columns
    
    # Filter the DataFrame for testing for the current fold
    df_test = df_dedup_v1[df_dedup_v1[f'fold_{current_fold}'] == 'test']
    df_test.reset_index(drop=True, inplace=True)
    df_test = df_test[used_columns] # Select only the used columns
    
    print(f' Train: {len(df_train)}')
    print(f' Val: {len(df_val)}')
    print(f' Test: {len(df_test)}')
    print()
```
