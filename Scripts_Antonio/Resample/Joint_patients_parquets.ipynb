```python
import os
import glob
import warnings
import pandas as pd

# Carpeta base de DiaTrend
base_folder = 'c:/Users/Ciro/C/_UGR_PDI/RELIEF-T1D/backup/Antonio/parquets/replace/'

# Versiones a procesar (subcarpetas v1…v4)
versions = ['v4']

# Silenciar warnings de tipos mixtos en pandas
warnings.filterwarnings('ignore', category=pd.errors.DtypeWarning)

for v in versions:
    folder = os.path.join(base_folder, v)
    if not os.path.isdir(folder):
        print(f"No existe la carpeta {folder}, la salto.")
        continue

    # 1) Concatenar resampled_{patient_id}.parquet
    resampled_pattern = os.path.join(folder, 'resampled_*.parquet')
    resampled_files = glob.glob(resampled_pattern)
    if resampled_files:
        dfs = [pd.read_parquet(f) for f in resampled_files]
        df_resampled = pd.concat(dfs, ignore_index=True)
        df_resampled['Patient_ID'] = df_resampled['Patient_ID'].astype(int)
        df_resampled = df_resampled.sort_values(
            by=['Patient_ID'],
            ignore_index=True,
            kind='mergesort'
        )
        out_res = os.path.join(folder, f'Glucose_measurements_{v}.parquet')
        df_resampled.to_parquet(out_res, index=False)
        print(f"Concatenados {len(resampled_files)} archivos resampled en {out_res}")
    else:
        print(f"No se encontraron archivos resampled en {folder}")
```
