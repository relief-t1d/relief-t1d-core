```python
import os
import glob
import warnings
import pandas as pd

# Carpeta base de DiaTrend
base_folder = 'c:/Users/Ciro/C/_UGR_PDI/RELIEF-T1D/backup/Antonio/parquets/t1diabetesgranada/v4'


# Silenciar warnings de tipos mixtos en pandas
warnings.filterwarnings('ignore', category=pd.errors.DtypeWarning)

if not os.path.isdir(base_folder):
    print(f"No existe la carpeta {base_folder}, la salto.")
    
# 1) Concatenar resampled_{patient_id}.parquet
resampled_pattern = os.path.join(base_folder, 'resampled_*.parquet')
resampled_files = glob.glob(resampled_pattern)

dfs = [pd.read_parquet(f) for f in resampled_files]
df_resampled = pd.concat(dfs, ignore_index=True)
# df_resampled['Patient_ID'] = df_resampled['Patient_ID'].astype(int)
# df_resampled = df_resampled.sort_values(
#     by=['Patient_ID'],
#     ignore_index=True,
#     kind='mergesort'
# )
out_res = os.path.join(base_folder, f'Glucose_measurements_v4.parquet')
```

```python
df_resampled.sort_values(by=['Patient_ID', 'Measurement_date', 'Measurement_time', '5min', '15min', 'Measurement'], inplace=True, ignore_index=True)
df_resampled
```
