```python
## ALL patients dataset
df_antonio_complete = pd.read_parquet('c:/Users/Ciro/C/_UGR_PDI/RELIEF-T1D/backup/Antonio/parquets/replace/Glucose_measurements_v4_TEMP.parquet')
```

```python
## Selected patients dataset
df_antonio_selected = pd.read_parquet('c:/Users/Ciro/C/_UGR_PDI/RELIEF-T1D/backup/Antonio/parquets/replace/Glucose_measurements_antonio_FINAL.parquet')
```

```python
# Selected patient info
df_antonio_patient_info = pd.read_parquet('c:/Users/Ciro/C/_UGR_PDI/RELIEF-T1D/backup/Antonio/parquets/replace/Patient_info.parquet')
```

```python
df_antonio_patient_info
```

```python
complete_antonio_patient_list = df_antonio_complete['Patient_ID'].unique()
selected_antonio_patient_list = df_antonio_selected['Patient_ID'].unique()
```

### Exclude patients present in complete but not in selected

```python
s_complete_patient_list = pd.Series(complete_antonio_patient_list)
s_complete_patient_list[~s_complete_patient_list.isin(selected_antonio_patient_list)]
```

```python
selected_antonio_patient_list
```

### Filter complete dataset to only include selected patients

```python
df_antonio_complete_filtered = df_antonio_complete[df_antonio_complete['Patient_ID'].isin(selected_antonio_patient_list)]
df_antonio_complete_filtered.reset_index(drop=True, inplace=True)
df_antonio_complete_filtered.info()
```

### Sort both datasets by Patient_ID, Measurement_date, Measurement_time, 5min, 15min, Measurement

```python
df_antonio_complete_filtered.sort_values(by=['Patient_ID', 'Measurement_date', 'Measurement_time', '5min', '15min', 'Measurement'], inplace=True, ignore_index=True)
df_antonio_selected.sort_values(by=['Patient_ID', 'Measurement_date', 'Measurement_time', '5min', '15min', 'Measurement'], inplace=True, ignore_index=True)
```

### Compare both datasets

```python
df_antonio_complete_filtered.compare(df_antonio_selected)
```

Both datasets are identical for the selected patients.
